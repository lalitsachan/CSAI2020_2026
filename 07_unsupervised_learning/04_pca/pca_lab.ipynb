# Hands-On Lab: Principal Component Analysis (PCA)
Accompanies **Chapter 22: Principal Component Analysis (PCA)**.

In this lab, we:
1. Implement PCA from scratch using Eigendecomposition of the sample covariance matrix.
2. Verify exact variance ratios against `scikit-learn`'s `PCA`.
3. Construct a **Scree Plot** and cumulative explained variance curve.
4. Project high-dimensional data into 2D for exploratory analysis.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_wine

data = load_wine()
X_scaled = StandardScaler().fit_transform(data.data)

sk_pca = PCA().fit(X_scaled)
cum_evr = np.cumsum(sk_pca.explained_variance_ratio_)

fig, ax1 = plt.subplots(figsize=(9, 5))
ax1.bar(range(1, 14), sk_pca.explained_variance_ratio_, alpha=0.6, color='steelblue', label='Individual')
ax1.set_xlabel('Principal Component Index ($j$)', fontsize=11)
ax1.set_ylabel('Variance Ratio', color='steelblue', fontsize=11)

ax2 = ax1.twinx()
ax2.plot(range(1, 14), cum_evr, color='crimson', marker='o', linewidth=2, label='Cumulative')
ax2.axhline(0.85, color='gray', linestyle='--', label='85% Threshold')
ax2.set_ylabel('Cumulative Ratio', color='crimson', fontsize=11)

plt.title('PCA Scree Plot & Cumulative Explained Variance', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)
plt.show()

Z_2d = sk_pca.transform(X_scaled)[:, :2]
plt.figure(figsize=(8, 6))
plt.scatter(Z_2d[:, 0], Z_2d[:, 1], c=data.target, cmap='viridis', edgecolors='k', s=60, alpha=0.8)
plt.xlabel('PC 1', fontsize=11)
plt.ylabel('PC 2', fontsize=11)
plt.title('2D Principal Component Projection of Wine Dataset', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()
```
