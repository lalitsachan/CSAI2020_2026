# Hands-On Lab: Density-Based Clustering (DBSCAN)
Accompanies **Chapter 21: DBSCAN & Density-Based Clustering**.

In this lab, we:
1. Demonstrate K-Means failure on non-convex geometric shapes (moons and concentric rings).
2. Apply DBSCAN to discover arbitrary geometries and automatically isolate noise.
3. Use the $k$-distance graph knee heuristic to select $\\epsilon$.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN, KMeans
from sklearn.datasets import make_moons
from sklearn.neighbors import NearestNeighbors

X, _ = make_moons(n_samples=400, noise=0.08, random_state=42)

km_labels = KMeans(n_clusters=2, random_state=42, n_init=10).fit_predict(X)
db_labels = DBSCAN(eps=0.2, min_samples=5).fit_predict(X)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
ax1.scatter(X[:, 0], X[:, 1], c=km_labels, cmap='viridis', edgecolors='k', s=35)
ax1.set_title('K-Means Failure on Non-Convex Geometry', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)

colors = ['red' if l == -1 else plt.cm.viridis(l / max(1, db_labels.max())) for l in db_labels]
ax2.scatter(X[:, 0], X[:, 1], c=colors, edgecolors='k', s=35)
ax2.set_title('DBSCAN Success: Non-Convex Clusters & Noise Filtered', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
