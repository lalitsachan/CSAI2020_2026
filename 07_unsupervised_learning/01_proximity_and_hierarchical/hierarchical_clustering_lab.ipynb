# Hands-On Lab: Hierarchical Clustering & Linkage Analysis
Accompanies **Chapter 19: Distance Matrices & Hierarchical Clustering**.

In this lab, we:
1. Compute pairwise distance matrices.
2. Generate Dendrograms using SciPy (`single`, `complete`, `average`, `ward`).
3. Apply horizontal cutting thresholds to extract discrete cluster labels.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.datasets import make_blobs

X, _ = make_blobs(n_samples=50, centers=3, cluster_std=0.8, random_state=42)
Z_ward = linkage(X, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(Z_ward, truncate_mode='level', p=5)
plt.axhline(y=7.0, color='crimson', linestyle='--', label='Horizontal Cut Threshold (Height = 7.0)')
plt.xlabel('Sample Index or Cluster Size', fontsize=11)
plt.ylabel("Ward's Variance Distance (ESS)", fontsize=11)
plt.title('Hierarchical Agglomerative Dendrogram (Ward Linkage)', fontsize=12, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.show()

labels = fcluster(Z_ward, t=7.0, criterion='distance')
print(f"Extracted {len(np.unique(labels))} discrete clusters from cutting at height 7.0")
```
