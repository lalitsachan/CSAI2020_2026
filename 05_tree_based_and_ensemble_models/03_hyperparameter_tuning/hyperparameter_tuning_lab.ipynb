# Hands-On Lab: Hyperparameter Tuning (Grid vs Randomized Search)
Accompanies **Chapter 15: Cross-Validation & Hyperparameter Tuning**.

In this lab, we:
1. Formulate hyperparameter search spaces for tree ensembles.
2. Benchmark `GridSearchCV` vs `RandomizedSearchCV` on execution time and score.
3. Validate the **Bergstra-Bengio Theorem** on efficiency in high dimensions.

```python
import time
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.datasets import make_classification
from scipy.stats import randint

X, y = make_classification(n_samples=1000, n_features=15, n_informative=8, random_state=42)

param_dist = {
    'n_estimators': randint(50, 200),
    'max_depth': randint(3, 15),
    'min_samples_split': randint(2, 11),
    'min_samples_leaf': randint(1, 10)
}

start = time.time()
rs = RandomizedSearchCV(RandomForestClassifier(random_state=42), param_distributions=param_dist, n_iter=25, cv=3, random_state=42, n_jobs=-1)
rs.fit(X, y)
t_rs = time.time() - start

print(f"RandomizedSearchCV Time: {t_rs:.2f} s | Best CV Score: {rs.best_score_:.4f}")
print("Best Parameters:", rs.best_params_)
```
