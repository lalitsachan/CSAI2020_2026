# Hands-On Lab: Decision Trees & Pruning Mechanics
Accompanies **Chapter 13: Decision Trees from Scratch**.

In this lab, we:
1. Visualize recursive partitioning tree structures with `plot_tree`.
2. Inspect Gini impurity calculations at parent vs child nodes.
3. Evaluate Cost-Complexity Pruning (`ccp_alpha`) to prevent overfitting.

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import load_iris

data = load_iris()
X, y = data.data[:, :2], data.target

dt = DecisionTreeClassifier(max_depth=3, criterion='gini', random_state=42).fit(X, y)

plt.figure(figsize=(12, 7))
plot_tree(dt, feature_names=data.feature_names[:2], class_names=data.target_names, filled=True, rounded=True, fontsize=10)
plt.title('Decision Tree Visualization (Max Depth = 3)', fontsize=13, fontweight='bold')
plt.show()
```
