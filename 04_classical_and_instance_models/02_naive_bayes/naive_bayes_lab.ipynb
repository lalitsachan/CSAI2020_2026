# Hands-On Lab: Text Classification with Naive Bayes & TF-IDF
Accompanies **Chapter 11: Naive Bayes Classification**.

In this lab, we:
1. Load consumer complaints from `Consumer_Complaints_train.csv.zip`.
2. Clean text, build bag-of-words / TF-IDF representations.
3. Train `MultinomialNB` with Laplace smoothing.
4. Evaluate precision, recall, and top predictive tokens per complaint category.

```python
import os
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

zip_path = os.path.abspath('../../data/Consumer_Complaints_train.csv.zip')
df = pd.read_csv(zip_path, compression='zip')

narrative_col = 'Consumer complaint narrative'
target_col = 'Product'

df_clean = df[[narrative_col, target_col]].dropna().sample(n=5000, random_state=42)
print("Class breakdown:\n", df_clean[target_col].value_counts().head(5))

X_train_raw, X_test_raw, y_train, y_test = train_test_split(df_clean[narrative_col], df_clean[target_col], test_size=0.25, random_state=42)

tfidf = TfidfVectorizer(max_features=3000, stop_words='english')
X_train_vec = tfidf.fit_transform(X_train_raw)
X_test_vec = tfidf.transform(X_test_raw)

nb_model = MultinomialNB(alpha=1.0).fit(X_train_vec, y_train)
y_pred = nb_model.predict(X_test_vec)

print("\nNaive Bayes Classification Report:\n")
print(classification_report(y_test, y_pred))
```
