# Hands-On Lab: Regularization (Ridge, Lasso & ElasticNet)
Accompanies **Chapter 8: Regularization and the Bias-Variance Tradeoff**.

In this lab, we:
1. Demonstrate the geometric mechanism of $L_1$ sparsity vs $L_2$ shrinkage.
2. Trace regularization coefficient paths across varying values of $\alpha$.
3. Tune optimal penalties using `RidgeCV` and `LassoCV`.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge, Lasso, ElasticNet, RidgeCV, LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_regression

X, y, true_coef = make_regression(n_samples=150, n_features=20, n_informative=5, noise=15.0, coef=True, random_state=42)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

alphas = np.logspace(-2, 3, 100)
ridge_coefs = []
lasso_coefs = []

for a in alphas:
    r = Ridge(alpha=a).fit(X_scaled, y)
    l = Lasso(alpha=a, max_iter=2000).fit(X_scaled, y)
    ridge_coefs.append(r.coef_)
    lasso_coefs.append(l.coef_)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(alphas, ridge_coefs)
ax1.set_xscale('log')
ax1.set_xlabel('Regularization Strength $\\alpha$', fontsize=11)
ax1.set_ylabel('Coefficients ($w_j$)', fontsize=11)
ax1.set_title('Ridge ($L_2$) Shrinkage Path (Smooth Decay)', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)

ax2.plot(alphas, lasso_coefs)
ax2.set_xscale('log')
ax2.set_xlabel('Regularization Strength $\\alpha$', fontsize=11)
ax2.set_ylabel('Coefficients ($w_j$)', fontsize=11)
ax2.set_title('Lasso ($L_1$) Feature Selection Path (Sparse Zeros)', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

lasso_cv = LassoCV(cv=5, random_state=42).fit(X_scaled, y)
print(f"Optimal Alpha chosen by CV: {lasso_cv.alpha_:.4f}")
print(f"Features retained by Lasso: {np.sum(lasso_cv.coef_ != 0)} out of 20")
```
